```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv()

client = ChatOpenAI(base_url="https://api.deepinfra.com/v1/openai", api_key=os.getenv('DEEPINFRA_API_TOKEN'), model="deepseek-ai/DeepSeek-V4-Pro", reasoning_effort="high")
```

```python
response = client.invoke('What is AI? Answer in like 20 words')
```

```python
# Safely extract reasoning content from the metadata dictionary
choices = response.response_metadata.get("choices", [])
if choices:
    reasoning_content = choices[0].get("message", {}).get("reasoning_content")
    if reasoning_content:
        print("=== REASONING CONTENT ===")
        print(reasoning_content)
        print("=========================\n")
```
